## C.5 Where to put the PLFS files

```python
plfs = Path("data/plfs")
expected = ["perv1.csv", "hhv1.csv", "perrv.csv", "hhrv.csv"]
if plfs.exists():
    for name in expected:
        f = plfs / name
        print(f"  {name:10s}", f"{f.stat().st_size / 1e6:6.0f} MB" if f.exists()
              else "missing")
    print("  layout workbook:", [f.name for f in plfs.glob("Data_Layout*.xlsx")])
else:
    print("data/plfs not found: see the steps above (Chapters 7, 8, 14-20 need it).")
```

## C.6 Working on a laptop with little memory

```python
import os

import pandas as pd

for var in ("OPENBLAS_NUM_THREADS", "OMP_NUM_THREADS", "MKL_NUM_THREADS"):
    print(f"{var:21s} {os.environ.get(var, '(not set)')}")

demo = pd.DataFrame({"code": [1, 2, 3] * 100_000})
print(f"int64 codes:   {demo.memory_usage(deep=True).sum() / 1e6:.1f} MB")
print(f"int8 codes:    {demo.astype('int8').memory_usage(deep=True).sum() / 1e6:.1f} MB")
print(f"as text:       {demo.astype(str).memory_usage(deep=True).sum() / 1e6:.1f} MB")
del demo
```
